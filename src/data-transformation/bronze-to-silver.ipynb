```python
# Bronze to Silver transformation
from pyspark.sql.functions import col
from pyspark.sql.types import *

# Get all tables from bronze
tables = dbutils.fs.ls("/mnt/creditrisk/bronze")

for table in tables:
    table_name = table.name.replace("/", "")
    print(f"Processing table: {table_name}")
    
    # Read parquet from bronze
    df = spark.read.format("parquet").load(f"/mnt/creditrisk/bronze/{table_name}")
    
    # Apply attribute type changes
    # Convert all string columns that should be numeric
    for field in df.schema.fields:
        if isinstance(field.dataType, StringType):
            try:
                df = df.withColumn(field.name, col(field.name).cast(DoubleType()))
            except:
                pass
    
    # Write to silver
    df.write.format("parquet") \
        .mode("overwrite") \
        .save(f"/mnt/creditrisk/silver/{table_name}")
    
    print(f"✅ {table_name} written to silver")

print("Bronze to Silver transformation complete!")
```
